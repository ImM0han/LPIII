```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def function(x):
    return (x + 3)**2

def derivative(x):
    return 2 * (x + 3)
```

```python
x = 2
learning_rate = 0.1
iterations = 100

for i in range(iterations):
    x = x - learning_rate * derivative(x)

print("Local Minimum x:", x)
print("Minimum y:", function(x))
```

```python
x_values = np.linspace(-6, 4, 100)
y_values = function(x_values)

plt.plot(x_values, y_values)
plt.scatter(x, function(x), color='red')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Gradient Descent')
plt.show()
```
